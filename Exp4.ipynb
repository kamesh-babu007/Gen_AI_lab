```python
# ============================================================
# CLEAN INSTALLATION FOR COLAB
# ============================================================

# Remove conflicting packages
!pip uninstall -y -q torch torchvision torchaudio transformers

# Install a known compatible PyTorch combination
!pip install -q torch==2.11.0 torchvision==0.26.0 torchaudio==2.11.0

# Install Transformers 4.x
# This version supports both summarization and question-answering
!pip install -q transformers==4.57.1 sentencepiece


# ============================================================
# IMPORTS
# ============================================================

import torch
from transformers import pipeline


# ============================================================
# CHECK INSTALLATION
# ============================================================

print("PyTorch version:", torch.__version__)

try:
    import torchvision
    print("Torchvision version:", torchvision.__version__)
except Exception as e:
    print("Torchvision check:", e)

print("CUDA available:", torch.cuda.is_available())


# ============================================================
# DEVICE
# ============================================================

device = 0 if torch.cuda.is_available() else -1

if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))
else:
    print("Running on CPU")


# ============================================================
# ARTICLE
# ============================================================

article = """
Generative AI refers to a class of artificial intelligence models capable of
producing new content such as text, images, audio, and video. Large Language
Models (LLMs) such as GPT and LLaMA are trained on massive text corpora and
can perform a wide range of natural language tasks including translation,
summarization, and question answering. These models are increasingly being
deployed in industry applications ranging from customer support to software
development, transforming how humans interact with machines.
"""


# ============================================================
# TEXT SUMMARIZATION
# ============================================================

print("\nLoading summarization model...")

summarizer = pipeline(
    "summarization",
    model="facebook/bart-large-cnn",
    device=device
)

summary = summarizer(
    article,
    max_length=45,
    min_length=20,
    do_sample=False
)


print("\n================================")
print("       TEXT SUMMARIZATION")
print("================================")

print("\nOriginal Article:")
print(article)

print("\nSummary:")
print(summary[0]["summary_text"])


# ============================================================
# QUESTION ANSWERING
# ============================================================

print("\nLoading question answering model...")

qa = pipeline(
    "question-answering",
    model="distilbert-base-cased-distilled-squad",
    device=device
)

question = "What are Large Language Models trained on?"

answer = qa(
    question=question,
    context=article
)


print("\n================================")
print("       QUESTION ANSWERING")
print("================================")

print("\nQuestion:", question)
print("Answer:", answer["answer"])
print("Confidence:", round(answer["score"], 3))
```
